# Question 1 — Definitions

### a. String
A string is a sequence of characters (symbols) from an alphabet.  
**DNA Example:** `ATCGGTA`



### b. Prefix
A prefix is the starting part of a string.  
**DNA Example:** For `ATCGGTA`, the prefix of length 3 is `ATC`.



### c. Suffix
A suffix is the ending part of a string.  
**DNA Example:** For `ATCGGTA`, the suffix of length 3 is `GTA`.



### d. Substring
A substring is a continuous sequence of characters within a string.  
**DNA Example:** For `ATCGGTA`, `CGG` is a substring.



### e. Pattern
A pattern is a smaller string we search for within a larger string.  
**DNA Example:** Pattern `CGG` inside `ATCGGTA`.



### f. Alphabet
An alphabet is the set of symbols used to form strings.  
**DNA Example:** DNA alphabet = `{A, T, C, G}`


### g. String matching
String matching is the process of finding occurrences of a pattern within a string.  
**DNA Example:** Searching for `CGG` in `ATCGGTA` → found at position 3.

# Question 4 — KMP Algorithm: LPS Array

Pattern: `ABABCABAB`



### Step-by-step construction
The LPS (Longest Proper Prefix which is also Suffix) array stores, for each index, the length of the longest prefix of the pattern that is also a suffix ending at that position.



### Completed Table

| Index | Character | LPS |
|-------|-----------|-----|
| 0     | A         | 0   |
| 1     | B         | 0   |
| 2     | A         | 1   |
| 3     | B         | 2   |
| 4     | C         | 0   |
| 5     | A         | 1   |
| 6     | B         | 2   |
| 7     | A         | 3   |
| 8     | B         | 4   |



### Explanation
- At index 0 (`A`): no proper prefix/suffix → LPS = 0  
- At index 2 (`A`): prefix `A` matches suffix `A` → LPS = 1  
- At index 3 (`B`): prefix `AB` matches suffix `AB` → LPS = 2  
- At index 4 (`C`): mismatch, reset → LPS = 0  
- At index 5 (`A`): prefix `A` matches suffix `A` → LPS = 1  
- At index 6 (`B`): prefix `AB` matches suffix `AB` → LPS = 2  
- At index 7 (`A`): prefix `ABA` matches suffix `ABA` → LPS = 3  
- At index 8 (`B`): prefix `ABAB` matches suffix `ABAB` → LPS = 4  

Thus, the final LPS array is:  
`[0, 0, 1, 2, 0, 1, 2, 3, 4]`
